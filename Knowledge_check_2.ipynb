1.   Consider two segments of the text. Compute the following probabilitiess of a bigram language model whose probabilities were estimated only on these two text segments:



*   P(, ∣ cloth)
*   P( cloth ∣ ,)
*   P( history ∣ culture)
*   P( history ∣ ,)







```python
import re
from collections import Counter, defaultdict
import pandas as pd
```

```python
text1 = """
the vibrant colors and intricate patterns of kente cloth , a symbol of
ghanaian royalty and prestige , tell stories of culture , history , and social
status . woven on narrow looms by skilled artisans , each strip of kente is
a testament to patience and artistry.
"""

text2 = """
bogolanfini , or mud cloth , from mali , is a textile steeped in tradition and
symbolism . its distinctive patterns , created using fermented mud and
natural dyes , tell stories of malian culture , history , and beliefs . the
process of creating bogolanfini is as rich and complex as the designs
themselves .
"""

# Separate punctuation and remove extra whitespace
def clean_text(text):
    text = re.sub(r"([.,])", r" \1 ", text)
    return " ".join(text.split())

dataset = [clean_text(text1), clean_text(text2)]
```

```python
# 2. TOKENIZATION (FROM THE NOTEBOOK)

def space_tokenize(text: str) -> list[str]:
    return text.split(" ")


# 3. GENERATE N-GRAMS (ACTIVITY 1)

def generate_ngrams(text: str, n: int) -> list[tuple[str]]:
    tokens = space_tokenize(text)
    ngrams = []

    for i in range(len(tokens) - n + 1):
        ngram = tuple(tokens[i:i+n])
        ngrams.append(ngram)

    return ngrams
```

```python
# 4. COUNT N-GRAMS (ACTIVITY 2)

def get_ngram_counts(dataset: list[str], n: int) -> dict[str, Counter]:
    ngram_counts = defaultdict(Counter)

    for paragraph in dataset:
        for ngram in generate_ngrams(paragraph, n):
            context = " ".join(ngram[:-1])
            next_token = ngram[-1]

            ngram_counts[context][next_token] += 1

    return dict(ngram_counts)


# 5. BUILD THE MODEL (ACTIVITY 3)

def build_ngram_model(dataset: list[str], n: int) -> dict:
    ngram_model = {}
    ngram_counts = get_ngram_counts(dataset, n)

    for context, next_tokens in ngram_counts.items():
        total = sum(next_tokens.values())
        ngram_model[context] = {}

        for token, count in next_tokens.items():
            ngram_model[context][token] = count / total

    return ngram_model


# 6. COMPUTE THE REQUESTED PROBABILITIES

bigram_model = build_ngram_model(dataset, n=2)

print("P(, | cloth) =", bigram_model["cloth"].get(",", 0))
print("P(cloth | ,) =", bigram_model[","].get("cloth", 0))
print("P(history | culture) =", bigram_model["culture"].get("history", 0))
print("P(history | ,) =", bigram_model[","].get("history", 0))
```
